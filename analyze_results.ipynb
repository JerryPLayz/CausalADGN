```python
from cadgn_statistics import run_analysis, load_variant_results
from pathlib import Path

do_models = [
    "Qwen/Qwen3-4B",
    "meta-llama/Llama-3.2-3B-Instruct",
    "Qwen/Qwen3-8B",
    "meta-llama/Llama-3.1-8B-Instruct"
]
model_ablations = ["cadgn", "adgn", "gcn", "gat", "dec"]


base_path = Path("./ablate_results/samples")

def get_filepath(abl, modelid, mmd_int, epochs="0040"):
    safe_mdl_name = modelid.replace("/", "__")
    test_path = base_path / f"{abl}_{safe_mdl_name}_mmd-{mmd_int}_{epochs}_per_sample.json"

    if not test_path.exists():
        print(f"Could not find file with path elements: ({abl}, {modelid} {mmd_int}) for epochs val `{epochs}`")
        return None
    return test_path

# (abl, modelid, mmd_int) = k
samples_dict ={(abl, mdl, int(mmdval)): get_filepath(abl, mdl, int(mmdval)) for mmdval in [0.0, 1.0] for abl in model_ablations for mdl in do_models}


baselines_to_add = {("baseline", mdl, int(mmdval)): get_filepath("baseline", mdl, 0, epochs="0150") for mmdval in [0.0, 1.0] for mdl in do_models}

# an unfortunate limit of square matrices.... (we'll ignore mmdval=1 later, as the baselines don't change over this)
for k,v in baselines_to_add.items():
    samples_dict[k] = v

print(samples_dict)
```

```python
variant_results = load_variant_results(samples_dict, 40, None)

print(len(variant_results))
```

```python
analysis = run_analysis(variant_results, num_cpus=24)  #threads
s1 = analysis.summary()
print(s1)
```

```python
import pandas as pd
def manual_to_dataframe(self) -> pd.DataFrame:
    return pd.DataFrame([
        {
            "variant_a": r.variant_a,
            "variant_b": r.variant_b,
            "test": r.test,
            "metric": r.metric,
            "statistic": r.statistic,
            "p_value": r.p_value,
            "p_value_corr": r.p_value_corr,
            "reject_h0": r.reject_h0,
            "direction": r.direction,
            "direction_int": r.direction_int,
            "n_samples": r.n_samples,
            "notes": r.notes,
        }
        for r in self.results
    ])

df = manual_to_dataframe(analysis)

sig = df[df["reject_h0"] == True]

lines = [
    f"Analysis Report (alpha={analysis.alpha}, Holm-Bonferroni corrected)",
    f"\tMcNemar tests:         {analysis.n_tests_mcnemar}",
    f"\tStuart-Maxwell tests:  {analysis.n_tests_stuart_maxwell}",
    f"\tASO tests:             {analysis.n_tests_aso} (per continuous metric)",
    f"\tSignificant Results:   {len(sig)}",
    "",
]

if sig.empty:
    lines.append(f"\t>>No significant results after correction.")
else:
    for _, row in sig.iterrows():
        lines.append(
            f" [{row['test']}] {row['variant_a']} vs {row['variant_b']} "
            f"| {row['metric']} | stat={row['statistic']:.4f} "
            f"| p_corr={row['p_value_corr']:.4f} "
            f"| {row['direction']}"
        )

summary = "\n".join(lines)


print(summary)
with open("./ablate_results/statistics_summary.txt", "w") as f:
    f.write(summary + "\n")

df.to_csv("./ablate_results/statistics_flat_all.csv", index=False)
```

```python
analysis.save("./ablate_results/", "all")
print("Saved Successfully")
with open("./ablate_results/statistics_summary.txt", "w") as f:
    f.write(analysis.summary() + "\n")
```
